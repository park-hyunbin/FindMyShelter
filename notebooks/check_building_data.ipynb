# Find My Shelter — QGIS 편집본 확인

편집은 QGIS의 data/working/opo_300m/buildings.shp에서 진행하세요. 이 노트북은 읽기 전용 점검용입니다.

```python
from pathlib import Path
import geopandas as gpd
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data/sample.json").exists())
source = ROOT / "data/working/opo_300m/buildings.shp"
buildings = gpd.read_file(source, engine="pyogrio", encoding="UTF-8")
print(len(buildings), buildings.crs)
display(buildings.drop(columns="geometry").head(20))
```

```python
display(buildings.loc[buildings.height_m.isna() | buildings.floors.isna()].drop(columns="geometry"))
```
